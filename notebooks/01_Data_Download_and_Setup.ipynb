```python
import pandas as pd
import glob

# 1. Define where the files are located in your Drive
benign_path = "/content/drive/MyDrive/EncryptedSDP/datasets/training/raw/benign-csvs/*.csv"
malic_path = "/content/drive/MyDrive/EncryptedSDP/datasets/training/raw/malic-csvs/*.csv"

print("Reading and combining training files from Google Drive... (This may take a few seconds)")

# 2. Load and tag all benign traffic
benign_files = glob.glob(benign_path)
benign_dfs = [pd.read_csv(f) for f in benign_files]
benign_data = pd.concat(benign_dfs, ignore_index=True)
benign_data['Threat_Label'] = 'Benign'

# 3. Load and tag all malicious attack traffic
malic_files = glob.glob(malic_path)
malic_dfs = [pd.read_csv(f) for f in malic_files]
malic_data = pd.concat(malic_dfs, ignore_index=True)
malic_data['Threat_Label'] = 'Malicious'

# 4. Combine everything into one master dataset
full_train_df = pd.concat([benign_data, malic_data], ignore_index=True)

# 5. Print the readiness report
print("\n--- MASTER TRAINING DATASET REPORT ---")
print(f"Total Benign Network Flows:    {len(benign_data):,}")
print(f"Total Malicious Network Flows: {len(malic_data):,}")
print(f"Total Combined Records:        {len(full_train_df):,}")
print(f"Total Features (Columns):      {full_train_df.shape[1]}")
```
